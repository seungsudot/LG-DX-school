```python
## header_info를 넣어서 오류 없애기 -> res 결과가 Response[400] 이 나오는 경우 이 방법 활용

# 개발자 모드의 console에다가 navigator.userAgent 입력한 결과
header_info = {"User-Agent" : "'Mozilla/5.0 (Linux; Android 6.0; Nexus 5 Build/MRA58N) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/131.0.0.0 Mobile Safari/537.36'"}

url = "https://www.melon.com/"
res = req.get(url, headers = header_info)
res
```

```python
## 검색 키워드에 따른 결과 예측하기 

search_url = "https://search.naver.com/search.naver?where=nexearch&sm=top_hty&fbm=0&ie=utf8&query=%EC%9D%8C%EC%8B%9D%EB%AC%BC%EC%93%B0%EB%A0%88%EA%B8%B0"

res = req.get(search_url)
res.text
```

```python
# 위 코드의 url을 보면 뒷 부분에 qurey = ~~~ 가 있음
# 이걸 활용해서 코드의 재사용성을 높임 


search_url = "https://search.naver.com/search.naver?where=nexearch&sm=top_hty&fbm=0&ie=utf8"

# params를 이용해서 결과 뽑음 
# 위 코드와 동일
res = req.get(search_url, params ={'query' : "음식물 쓰레기"})
res.text
```
