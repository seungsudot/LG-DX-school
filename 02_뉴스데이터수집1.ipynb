### 복습 css 선택자
1. 태그 선택자 : 태그 이름으로 요소를 선택
2. 클래스 선택자 : . 
3. 아이디 선택자 : #
4. 계측 선택자
   * 자식 선택자 : A>B (A가 부모, B 자손)
   * 자손 선택자 : A B (A가 부모, B가 자손)
   * 인접 형제 선택자 A+B (A는 형제 1, B는 형제 2)
   * 형제 선택자 : A~B 형태로 

```python
url = "https://n.news.naver.com/article/277/0005528627"
res = req.get(url)
res.text

# res.text를 lxml로 가져온다. 
html = bs(res.text , "lxml")
```

```python
#  #title_area > span는 개발자도구에서 copy selector를 하면된다. 
html.select_one("#title_area > span").text
```

```python
#   #dic_area가 id 선택자
html.select_one("#dic_area").text
```

### 여러 페이지의 내용을 수집해야 할 경우

* 눈에 보이지 않는 데이터 수집 필요 -> a태그의 속성값인 href
  ### 데이터 수집 순서
  1. 뉴스 페이지의 url 주소 수집 및 리스트로 저장
  2. 리스트와 반복문을 활용해서 각 뉴스 페이지 문서 요청 (request)
  3. 각 뉴스 페이지 문서를 변환하는 작업
  4. 변환된 문서 내 뉴스 제목, 내용을 수집한다. 

```python
news_selection_url = "https://news.naver.com/section/105"
res = req.get(news_selection_url)
html = bs(res.text, "lxml")
```

```python
# select(css선택자) : 선택자 조건에 해당하는 모든 html 요소를 하나의 리스트로 반환 


news_list = html.select("div.sa_text > a")
news_list 
```

```python
# get (태그의 속성 이름 ): html 요소의 속성값을 문자열로 반환 
news_list[0].get("href")
```

# 실습 news_list에 저장된 모든 a 요소의 속성값을 list에 저장하기

```python
a_list =[]
for i in news_list:
    a_list.append(i.get("href"))

a_list
```

```python
# 수집한 url 접근해서, title, content 모아보기 

for href in a_list:
    res = req.get(href)
    html = bs(res.text, "lxml")

    # title과 content 
    title = html.select_one("#title_area > span").text
    content = html.select_one("#dic_area").text.strip()

    print("뉴스 제목은", title)
    print("내용 : "+ content)
    print("\n" *20)
    
    
```

```python
# 수집한 url 접근해서, title, content 모아보기 

data_list =[]

for href in a_list:
    res = req.get(href)
    html = bs(res.text, "lxml")

    # title과 content 
    title = html.select_one("#title_area > span").text
    content = html.select_one("#dic_area").text.strip()

    data_list.append([title,content, href])

data_list
    
```

```python
# 저장하기 

import pandas as pd 

news_df = pd.DataFrame(data_list , columns =['뉴스제목', '내용', '페이지 주소'])
news_df
```

```python
news_df.to_csv("./뉴스.csv")
```
