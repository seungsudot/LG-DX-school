### ZDNET KOREA 사이트의 뉴스 제목, 주소를 수집하여 DataFrame으로 만들어보기

사이트 주소 :

1. url 주소의 pageno를 이용하여 1~20 페이지 문서 요청
2. 응답 받은 웹 문서를 변환하기
3. 변환된 웹 문서에서 뉴스의 제목, 주소(href) 수집 후 리스트로 저장
4. 리스트를 DataFrame으로 생성 후 저장

```python
import requests as req
import pandas as pd
import time

from bs4 import BeautifulSoup as bs
```

```python
news_selection_url = "https://search.zdnet.co.kr/?kwd=%EC%9D%8C%EC%8B%9D%EB%AC%BC"

res = req.get(news_selection_url)

html = bs(res.text, "lxml")
```

```python
# 연관된 정보를 가져올때는 a 태그 
news_list = html.select("div.assetText a")
news_list 

```

```python
a_list =[]
for i in news_list:
    a_list.append(i.get("href"))
a_list
```

```python
# 가져온 것의 제목, 주소 저장
data_list =[]

for href in a_list:
    res = req.get(href)
    html = bs(res.text, "lxml")

    # title과 href
    title = html.select_one("div.assetText > a > h3").text
    data_list.append([title, href])

```

```python
# 위 코드의 url을 보면 뒷 부분에 qurey = ~~~ 가 있음
# 이걸 활용해서 코드의 재사용성을 높임 
search_url = "https://search.zdnet.co.kr/"

all_df = pd.DataFrame(columns =['제목','주소'])

# params를 이용해서 결과 뽑음 
# 위 코드와 동일

for i in range (1,20):
    res = req.get(search_url, params ={'kwd' : "음식물", 'pageno':f'{i}'})
    html = bs(res.text, "lxml")

    # 뉴스 리스트 뽑아오기
    news_list = html.select("div.assetText > a")


    # 그걸 바탕으로 a링크의 href 주소를 저장
    a_list =[]
    for h in news_list:
        a_list.append(h.get("href"))


    # 가져온 것의 제목, 주소 저장
    data_list =[]
    
    for href in a_list:
        res2 = req.get(href)
        html2 = bs(res2.text, "lxml")
    
        # title과 href
        title = html2.select_one("div.news_head > h1").text
        data_list.append([title, href])

    df = pd.DataFrame(data_list, columns =['제목', '주소'])
    all_df = pd.concat([all_df, df])
    
        
```
