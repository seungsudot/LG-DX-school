```python
# Selenium을 사용하기 위한 3가지 라이브러리

#1. 브라우저 실행 및 정보를 보관하는 역할
from selenium import webdriver as wb

#2. 요소를 찾기(선택)위한 도구를 제공하는 역할
from selenium.webdriver.common.by import By

#3. 컴퓨터 키보드의 값을 제공하는 역할 
from selenium.webdriver.common.keys import Keys
```

```python
# 1. 브라우저 실행 
# Chrome() 함수를 호출, 테스트 브라우저가 실행
# 한 번만 실행되도록 할 것 -> 호출할때마다 새로운 브라우저 실행되기 때문에

driver = wb.Chrome()
```

```python
# 2. 네이버 페이지 요청 
# 아까 띄운 크롬창에서 네이버로 접속됨
driver.get("https://www.naver.com/")
```

```python
# 3. 검색창에 "음식물" 키워드 보내기 

#3-1. 검색창 요소에 접근
query_input = driver.find_element(By.ID, 'query')
#3-2. 접근한 요소에 키워드 전송
query_input.send_keys("음식물")
# 엔터
query_input.send_keys(Keys.ENTER)
```

```python
# 참고 ) 뒤로 가는 법
driver.back()
```

```python
# 5. 더보기 버튼 클릭하기 

btn= driver.find_element(By.CSS_SELECTOR, "#lnb > div.lnb_group > div > div.lnb_nav_area._nav_area_root > div > div.api_btn_page._nav_btn_root > div.btn_next._next > a")
btn.click()
```

```python
# 6. 뉴스 탭 클릭하기

news_btn = driver.find_element(By.XPATH, '//*[@id="lnb"]/div[1]/div/div[1]/div/div[1]/div[9]')
news_btn.click()
```

```python
# 드라이버 종료 
driver.close()
```
