```python
from kiwipiepy import Kiwi

kiwi = Kiwi()
```

```python
kiwi.tokenize("오늘 수업은 정말 고되네요")
```

```python
#텍스트 파일 읽어오기 

f = open('빅카인즈 뉴스 데이터 1.txt', 'r', encoding ='utf-8')
text = f.readline()
f.close()
```

```python
text
```

```python
# fom : 실제 단어
# tag : 단어의 품사
# start : 단어의 시작인덱스
# len : 단어의 길이
kiwi.tokenize(text)
```

```python
result = kiwi.tokenize(text)
```

```python
# 일반 명사, 고유 명사만 리스트에 저장해보기


# 속성으로 꺼내볼 수 있어요
print(result[0])
print(result[0].form)
print(result[0].tag)
```

```python
noun_list = []

for token in result:
    # if token.tag == 'NNG' or token.tag == 'NNP'
    if token.tag in ["NNG", "NNP"]:
        noun_list.append(token.form)
```

```python
noun_list
```

```python
# 단어의 빈도수를 카운팅해주는 도구

from collections import Counter
counter = Counter(noun_list)
counter
```

```python
# 텍스트 데이터 전처리
# 상위 100개의 단어만 추출
# 2글자 이상의 단어만 추출

top_100_noun_list = counter.most_common(100)
top_100_noun_list
```

```python
noun_list2 = []

for noun in top_100_noun_list:
    if len(noun[0]) >= 2:
        noun_list2.append(noun)
```

```python
len(noun_list2)
```

```python
from wordcloud import WordCloud #워드클라우드를 생성하기 위한 도구
import matplotlib.pyplot as plt

wc= WordCloud(
    # 폰트 설정 , 따로 안해주면 한글 인식 안돼...
    font_path = 'C:/Windows/Fonts/malgun.ttf',
    background_color ='white',
    width = 500,
    height = 500
).generate_from_frequencies(dict(noun_list2))

plt.imshow(wc)
plt.axis('off')
plt.show()
```
