```python
from kiwipiepy import Kiwi
kiwi = Kiwi()


result_list =[]


# 하나씩 불러와
for i in range (2004, 2025):
    read_df = pd.read_csv(f'./year/blog_{i}.csv')

    # df의 모든 text를 하나의 word로 변환
    combined_word = ' '.join(read_df.astype(str).values.flatten())
    #print("방법 1 결과:", combined_word)a
    
    
    # kiwi로 토큰단위 분리
    result = kiwi.tokenize(combined_word)
    
    
    # 일반명사, 고유명사만 리스트에 저장
    noun_list = []
    
    for token in result:
        if token.tag in ['NNG','NNP']:
            noun_list.append(token.form)
    
    
    # 단어의 빈도를 카운팅 해주는 도구
    
    from collections import Counter
    
    counter = Counter(noun_list)
    
    # 텍스트 데이터 전처리
    # 상위 100개의 단어만 추출
    top_100_noun_list = counter.most_common(200)
    
    # 2글자 이상의 단어만 추출
    noun_list2 = []
    
    for noun in top_100_noun_list:
        if len(noun[0]) > 1:
            noun_list2.append(noun)

    result_list.append([noun_list2])
```

```python
concat_all_df = pd.DataFrame()

for i in range(0,len(result_list)):
    print(f"==============={2004+i}=================")
    print(f"{2004+i}의 데이터입니다.")
    print(result_list[i])
    print("\n\n\n")

    # Flatten을 먼저 처리해줍니다. 
    flattened_data = [item for sublist in result_list[i] for item in sublist]
    df = pd.DataFrame(flattened_data, columns=['keyword', '빈도'])
    concat_all_df = pd.concat([concat_all_df,df ], axis =1)
```

```python
concat_all_df.to_excel("./result/blog_concat_df.xlsx", index = False)
```
